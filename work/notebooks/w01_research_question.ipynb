## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

url = url = "https://raw.githubusercontent.com/Tuborg49/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Number of rows:", len(df))
print("Average 90-day impressions:", round(df["impressions_90d"].mean(), 2))
print("Percentage of pages with declining trend:",
      round((df["trend_direction"] == "down").mean() * 100, 2))
```
